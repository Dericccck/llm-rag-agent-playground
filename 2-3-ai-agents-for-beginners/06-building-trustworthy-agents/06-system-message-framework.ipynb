# 说明
此代码提供了能够更轻松地为多个代理扩展创建系统消息的系统消息框架

原来的方法用AZURE掉openai大模型，下面写了一个掉千问大模型，提示词都是一样的，原来的方法在改写的方法下面，可以对比两个大模型输出的内容，可以看出还是有差距的

```python
# 改写方法，使用千问大模型创建系统消息框架
import os
from openai import OpenAI # 导入异步 OpenAI 客户端
from dotenv import load_dotenv # 导入 load_dotenv，用于从 .env 文件中加载环境变量。

load_dotenv()

# model_name = "qwen-max"

# # 创建 AsyncOpenAI 客户端实例
# client = OpenAI(
#     api_key=os.environ.get("DASHSCOPE_API_KEY"), 
#     base_url="https://dashscope.aliyuncs.com/compatible-mode/v1",
# )

model_name = "gpt-4.1-mini"
client = OpenAI(
    api_key=os.environ["GITHUB_TOKEN"],
    base_url="https://models.inference.ai.azure.com/"
)

role = "travel agent"
company = "contoso travel"
responsibility = "预订航班"

response = client.chat.completions.create(
    messages=[
        {"role": "system", "content": """你是创造人工智能代理助手的专家。
您将获得公司名称、角色、职责和其他信息
将用于为其提供系统提示的信息。
要创建系统提示，请尽可能具有描述性，并提供一个结构，使使用LLM的系统能够更好地理解AI助手的角色和职责。"""},
        {"role": "user", "content": f"你是{company}的{role}，负责{responsibility}。"},
    ],
    model=model_name,
    temperature=1.0,
    max_tokens=1000,
    top_p=1.0
)

print(response.choices[0].message.content)
```

### 下方是原来的代码

```python
import os
from azure.ai.inference import ChatCompletionsClient
from azure.ai.inference.models import SystemMessage, UserMessage
from azure.core.credentials import AzureKeyCredential
from dotenv import load_dotenv # 导入 load_dotenv，用于从 .env 文件中加载环境变量。

load_dotenv()
token = os.environ["GITHUB_TOKEN"]
endpoint = "https://models.inference.ai.azure.com"
```

```python
model_name = "gpt-4o"

client = ChatCompletionsClient(
    endpoint=endpoint,
    credential=AzureKeyCredential(token),
)
```

```python
role = "travel agent"
company = "contoso travel"
responsibility = "booking flights"
```

```python
response = client.complete(
    messages=[
        SystemMessage(content="""You are an expert at creating AI agent assistants. 
You will be provided a company name, role, responsibilities and other
information that you will use to provide a system prompt for.
To create the system prompt, be descriptive as possible and provide a structure that a system using an LLM can better understand the role and responsibilities of the AI assistant."""),
        UserMessage(content=f"You are {role} at {company} that is responsible for {responsibility}."),
    ],
    model=model_name,
    # Optional parameters
    temperature=1.,
    max_tokens=1000,
    top_p=1.
)

print(response.choices[0].message.content)
```
